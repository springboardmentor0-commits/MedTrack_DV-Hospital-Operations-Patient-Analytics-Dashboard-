# Master Dataset Creation

This notebook creates specialized master datasets from the cleaned Hospital Master Dataset.

Master Datasets:
- Doctor Master
- Billing Master
- Bed Master

```python
import pandas as pd
import numpy as np
import random
```

```python
hospital_master = pd.read_csv(
    "../data/02_processed_data/hospital_master_dataset.csv",
    parse_dates=["Admission_Date", "Discharge_Date"]
)
```

```python
hospital_master.info()
```

## Doctor Master Dataset

```python
doctor_master = hospital_master[
    [
        "Doctor_ID",
        "Doctor_Name",
        "Department_ID",
        "Department_Name",
        "Specialization",
        "Experience",
        "Patient_ID",
        "Name",
        "Age",
        "Gender",
        "Diagnosis",
        "Admission_ID",
        "Admission_Date",
        "Discharge_Date",
        "Length_of_Stay",
        "Status"
    ]
].copy()
```

```python
doctor_master.head()
```

```python
doctor_master.info()
```

```python
doctor_master.to_csv(
    "../data/02_processed_data/doctor_master.csv",
    index=False
)
```

## Billing Master Dataset

```python
billing_master = hospital_master[
    [
        "Bill_ID",
        "Admission_ID",
        "Patient_ID",
        "Name",
        "Age",
        "Gender",
        "Department_Name",
        "Doctor_Name",
        "Total",
        "Insurance",
        "Paid",
        "Pending",
        "Admission_Date",
        "Discharge_Date",
        "Length_of_Stay"
    ]
].copy()
```

```python
billing_master.head()
```

```python
billing_master.to_csv(
    "../data/02_processed_data/billing_master.csv",
    index=False
)
```
