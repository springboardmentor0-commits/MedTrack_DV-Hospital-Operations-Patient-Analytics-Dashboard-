# Bed Master Dataset Generation

## Objective

This notebook creates a **Bed Master Dataset** for the MedTrack project.

The dataset is generated using the **Hospital Master Dataset** and includes bed-related information such as bed ID, ward, room number, bed type, and bed status.

The generated dataset is linked using:
- Admission_ID
- Patient_ID
- Department_ID

The final dataset will be saved as **bed_master.csv** for hospital resource utilization and bed occupancy analysis.

```python
hospital_master = pd.read_csv(
    "../data/02_processed_data/hospital_master_dataset.csv",
    parse_dates=["Admission_Date", "Discharge_Date"]
)

hospital_master.head()
```

```python
ward_mapping = {
    "Cardiology": "Cardiac Ward",
    "Neurology": "Neuro Ward",
    "Orthopedics": "Orthopedic Ward",
    "Pediatrics": "Pediatric Ward",
    "Oncology": "Cancer Ward",
    "Emergency": "Emergency Ward",
    "General Medicine": "General Ward",
    "Dermatology": "Skin Care Ward",
    "Gynecology": "Maternity Ward",
    "Psychiatry": "Mental Health Ward"
}
```

```python
bed_types = [
    "General",
    "Semi-Private",
    "Private",
    "ICU"
]
```

```python
bed_status = [
    "Occupied",
    "Available",
    "Reserved",
    "Cleaning",
    "Maintenance"
]
```

```python
bed_master = pd.DataFrame()

bed_master["Bed_ID"] = [
    f"B{i:05d}" for i in range(1, len(hospital_master)+1)
]

bed_master["Admission_ID"] = hospital_master["Admission_ID"]

bed_master["Patient_ID"] = hospital_master["Patient_ID"]

bed_master["Department_ID"] = hospital_master["Department_ID"]

bed_master["Department_Name"] = hospital_master["Department_Name"]

bed_master["Ward"] = hospital_master["Department_Name"].apply(
    lambda x: ward_mapping.get(x, "General Ward")
)

bed_master["Floor"] = [
    random.randint(1,6)
    for _ in range(len(hospital_master))
]

bed_master["Room_Number"] = [
    random.randint(100,699)
    for _ in range(len(hospital_master))
]

bed_master["Bed_Number"] = [
    random.randint(1,4)
    for _ in range(len(hospital_master))
]

bed_master["Bed_Type"] = [
    random.choice(bed_types)
    for _ in range(len(hospital_master))
]

bed_master["Bed_Status"] = [
    random.choice(bed_status)
    for _ in range(len(hospital_master))
]

bed_master["Admission_Date"] = hospital_master["Admission_Date"]

bed_master["Discharge_Date"] = hospital_master["Discharge_Date"]

bed_master["Length_of_Stay"] = hospital_master["Length_of_Stay"]
```
