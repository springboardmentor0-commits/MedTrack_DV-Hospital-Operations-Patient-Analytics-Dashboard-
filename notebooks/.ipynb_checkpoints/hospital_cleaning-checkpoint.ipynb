```python
hospital.columns = (
    hospital.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("-", "_")
)
```

```python
text_cols = hospital.select_dtypes(include="object").columns

for col in text_cols:
    hospital[col] = hospital[col].str.strip()
```

```python
hospital["Stay_Category"] = pd.cut(
    hospital["Length_of_Stay_days"],
    bins=[0, 3, 7, 100],
    labels=["Short Stay", "Medium Stay", "Long Stay"]
)
```

```python
hospital["Wait_Category"] = pd.cut(
    hospital["Wait_Time_mins"],
    bins=[0, 15, 30, 1000],
    labels=["Low", "Medium", "High"]
)
```

```python
hospital["Bill_Category"] = pd.cut(
    hospital["Bill_Amount"],
    bins=[0, 10000, 30000, 1000000],
    labels=["Low", "Medium", "High"]
)
```

```python
hospital.to_csv("../data/hospital_cleaned.csv", index=False)
```

```python
datasets = {
    "Hospital": hospital,
    "Patients": patients,
    "Appointments": appointments,
    "Doctors": doctors,
    "Staff": staff,
    "Staff Schedule": staff_schedule,
    "Billing": billing,
    "Insurance": insurance,
    "Treatments": treatments,
    "Services": services,
    "Summary": summary
}

for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("Shape:", df.shape)
    print("Missing Values:")
    print(df.isnull().sum())
    print("Duplicate Rows:", df.duplicated().sum())
    print()
```

```python
def clean_column_names(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.replace(" ", "_")
        .str.replace("(", "", regex=False)
        .str.replace(")", "", regex=False)
        .str.replace("-", "_")
    )
    return df
```

```python
hospital = clean_column_names(hospital)
patients = clean_column_names(patients)
appointments = clean_column_names(appointments)
doctors = clean_column_names(doctors)
staff = clean_column_names(staff)
staff_schedule = clean_column_names(staff_schedule)
billing = clean_column_names(billing)
insurance = clean_column_names(insurance)
treatments = clean_column_names(treatments)
services = clean_column_names(services)
summary = clean_column_names(summary)
```

```python
hospital["Admission_Date"] = pd.to_datetime(hospital["Admission_Date"])
hospital["Discharge_Date"] = pd.to_datetime(hospital["Discharge_Date"])

patients["arrival_date"] = pd.to_datetime(patients["arrival_date"])
patients["departure_date"] = pd.to_datetime(patients["departure_date"])

appointments["appointment_date"] = pd.to_datetime(appointments["appointment_date"])

billing["bill_date"] = pd.to_datetime(billing["bill_date"])

treatments["treatment_date"] = pd.to_datetime(treatments["treatment_date"])
```
