## 4. Data Cleaning

Standardize column names across all datasets to maintain consistency.

```python
for name, df in datasets.items():

    df.columns = (
        df.columns
          .str.strip()
          .str.replace(" ", "_", regex=False)
          .str.replace("(", "", regex=False)
          .str.replace(")", "", regex=False)
          .str.replace("-", "_", regex=False)
    )

print(" Column names standardized.")
```

### Remove Leading and Trailing Spaces

```python
for name, df in datasets.items():

    text_cols = df.select_dtypes(include="object").columns

    for col in text_cols:
        df[col] = df[col].str.strip()

print("Extra spaces removed.")
```

### Remove Duplicate Records

```python
for name, df in datasets.items():

    before = len(df)

    df.drop_duplicates(inplace=True)

    after = len(df)

    print(f"{name}: {before-after} duplicate rows removed.")
```

### Convert Date Columns

```python
admissions["Admission_Date"] = pd.to_datetime(admissions["Admission_Date"])
admissions["Discharge_Date"] = pd.to_datetime(admissions["Discharge_Date"])

print(admissions.dtypes)
```

```python
for name, df in datasets.items():

    print("="*60)
    print(name)

    print("Missing Values :", df.isnull().sum().sum())
    print("Duplicate Rows :", df.duplicated().sum())
```

## 5. Feature Engineering

```python
admissions["Length_of_Stay"] = (
    admissions["Discharge_Date"] -
    admissions["Admission_Date"]
).dt.days
```

```python
admissions["Admission_Year"] = admissions["Admission_Date"].dt.year
```

```python
admissions["Admission_Month"] = admissions["Admission_Date"].dt.month_name()
```

```python
admissions["Admission_Quarter"] = admissions["Admission_Date"].dt.quarter
```

```python
admissions["Weekend_Admission"] = (
    admissions["Admission_Date"].dt.dayofweek >= 5
)
```

```python
admissions["Stay_Category"] = pd.cut(
    admissions["Length_of_Stay"],
    bins=[0,3,7,15,100],
    labels=["Short","Medium","Long","Extended"]
)
```

```python
patients["Age_Group"] = pd.cut(
    patients["Age"],
    bins=[0,18,35,60,100],
    labels=["Child","Young Adult","Adult","Senior"]
)
```

## 6. Data Integration

Merge all datasets using their verified primary and foreign keys to create a master hospital dataset.

```python
patient_master = patients.merge(
    admissions,
    on="Patient_ID",
    how="inner"
)

print("Shape:", patient_master.shape)
patient_master.head()
```

```python
print("Missing Values:", patient_master.isnull().sum().sum())
print("Duplicate Rows:", patient_master.duplicated().sum())
```

```python
patient_master = patient_master.merge(
    doctors,
    on="Doctor_ID",
    how="left"
)

print(patient_master.shape)
```

```python
print(patient_master.isnull().sum())
```

```python
patient_master = patient_master.merge(
    departments,
    on="Department_ID",
    how="left"
)

print(patient_master.shape)
```

```python
print(patient_master.columns.tolist())
```

```python
print(patient_master[["Department_ID_x", "Department_ID_y"]].head())

print(
    (patient_master["Department_ID_x"] == patient_master["Department_ID_y"]).all()
)
```

```python
patient_master.drop(columns=["Department_ID_y"], inplace=True)

patient_master.rename(
    columns={"Department_ID_x": "Department_ID"},
    inplace=True
)
```

```python
print(patient_master.columns.tolist())
```

```python
patient_master = patient_master.merge(
    departments,
    on="Department_ID",
    how="left",
    validate="many_to_one"
)

print(patient_master.shape)
```

```python
print(patient_master[[
    "Department_ID",
    "Department_Name"
]].head())
```

```python
patient_master = patient_master.merge(
    billing,
    on=["Admission_ID", "Patient_ID"],
    how="left"
)

print(patient_master.shape)
```

```python
patient_master = patient_master.merge(
    surgeries,
    on=["Admission_ID", "Patient_ID", "Doctor_ID"],
    how="left",
    suffixes=("", "_Surgery")
)

print(patient_master.shape)
```

```python
patient_master = patient_master.merge(
    medications,
    on="Admission_ID",
    how="left"
)

print(patient_master.shape)
```

```python
patient_master = patient_master.merge(
    lab_results,
    on="Admission_ID",
    how="left",
    suffixes=("", "_Lab")
)

print(patient_master.shape)
```

```python
# Admissions → Doctors
validate="many_to_one"

# Admissions → Departments
validate="many_to_one"

# Admissions → Billing
validate="one_to_one"

# Admissions → Surgeries
validate="one_to_one"

# Admissions → Medications
validate="one_to_one"

# Admissions → Lab Results
validate="one_to_one"
```
